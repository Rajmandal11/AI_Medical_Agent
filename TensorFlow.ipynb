```python
import tensorflow as tf
```

```python
mnist = tf.keras.datasets.mnist

(x_train, y_train),(x_test, y_test) = mnist.load_data()
x_train, x_test = x_train/255, x_test/255
```

```python
model = tf.keras.models.Sequential([tf.keras.layers.Flatten(input_shape=(28,28)),
                                    tf.keras.layers.Dense(128, activation = 'relu'),
                                    tf.keras.layers.Dropout(0.2),
                                    tf.keras.layers.Dense(10)])
```

```python
predications = model(x_train[:1]).numpy()
predications
```

```python
tf.nn.softmax(predications).numpy() 
```

```python
loss_nf = tf.keras.losses.SparseCategoricalCrossentropy(from_logits= True) # actuall predication
loss_nf(y_train[:1], predications).numpy()
```

```python
model.compile(optimizer='adam', loss = loss_nf, metrics = ['accuracy'])
model.fit(x_train, y_train, epochs = 5)
```

```python
model.evaluate(x_test, y_test, verbose = 2) # To check the eaccuracy and loss of model
```

```python
probabilaty_model = tf.keras.Sequential([model, tf.keras.layers.Softmax()]) # actuall predication of model
probabilaty_model(x_test[:3])
```

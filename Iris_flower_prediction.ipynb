```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier 
from sklearn.metrics import accuracy_score 
```

```python
data = pd.read_csv('Iris.csv')
data.head()
```

```python
x = data.iloc[:,:-1]
y = data["Species"]
```

```python
x_train, x_test, y_train, y_test = train_test_split(x,y, test_size = 0.2, random_state = 39)
```

```python
n_estimators_list = [20, 50, 100, 200, 500]

print("Random Forest Classifier Performance Comparison (n_estimators):") 
print("-" * 50) 
results = {} 
```

```python
for n_trees in n_estimators_list:
    rf_model = RandomForestClassifier(n_estimators=n_trees, random_state=42) 
    rf_model.fit(x_train, y_train)
    y_pred = rf_model.predict(x_test) 
    accuracy = accuracy_score(y_test, y_pred) 
    results[n_trees] = accuracy 
    print(f"Number of Trees: {n_trees:3d} | Accuracy: {accuracy:.4f}") 
    print("-" * 50) 
```

```python
best_n_trees = max(results, key=results.get) 
print(f"The best performance was achieved with {best_n_trees} trees, with an accuracy of {results[best_n_trees]:.4f}.") 
```

```python
plt.plot(list(results.keys()), list(results.values()), marker='o') 
plt.title('Accuracy vs. Number of Trees') 
plt.xlabel('Number of Trees (n_estimators)') 
plt.ylabel('Accuracy') 
plt.grid(True) 
plt.show() 
```

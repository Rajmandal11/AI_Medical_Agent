```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
```

```python
dataset = pd.read_csv('stock_data.csv')
```

```python
dataset.head()
```

```python
dataset.isnull().sum()
```

```python
dataset.drop('Unnamed: 0', axis = 1, inplace = True)
```

```python
sns.pairplot(data = dataset)
plt.show()
```

```python
plt.figure(figsize = (15,5))
plt.plot(dataset['Stock_1'])
plt.title('Stock_1')
plt.xlabel('predict')
plt.ylabel('Price in dollars')
plt.show()
```

```python
features = ['Stock_1', 'Stock_2', 'Stock_3', 'Stock_4', 'Stock_5']

plt.subplots(figsize=(20,10))

for i, col in enumerate(features):
  plt.subplot(2,3,i+1)
  sns.distplot(dataset[col])
plt.show()
```

```python
features = ['Stock_1', 'Stock_2', 'Stock_3', 'Stock_4', 'Stock_5']

plt.subplots(figsize=(20,10))

for i, col in enumerate(features):
  plt.subplot(2,3,i+1)
  sns.boxplot(dataset[col])
plt.show()
```

```python
wcss = []
for i in range (3,24):
    kn = KMeans(n_clusters  = i, init = 'k-means++')
    kn.fit(dataset)
    wcss.append(kn.inertia_)
```

```python
plt.figure(figsize = (15,5))
plt.plot([i for i in range (3,24)], wcss, marker = 'o')
plt.xticks([i for i in range (3,24)])
plt.xlabel('No. of Clusters')
plt.ylabel('WCSS')
plt.grid(axis = 'x')
plt.savefig('Cluster.jpg')
plt.show()
```

```python
kms = KMeans(n_clusters = 5)
dataset['predict'] =  kms.fit_predict(dataset)
```

```python
kms.fit_predict(dataset)
```

```python
dataset.head()
```

```python
silhouette_score(dataset, labels = kms.labels_)
```

```python
ss = []
no_c = [j for j in range(4,24)]
for i in range(4,24):
    km1 = KMeans(n_clusters = 5)
    km1.fit(dataset)
    ss.append(silhouette_score(dataset,km1.labels_))
```

```python
plt.plot(no_c,ss, marker = 'o')
plt.xticks(no_c)
plt.xlabel('No. of Clusters')
plt.ylabel('SS')
plt.grid(axis = 'x')
plt.show()
```

```python
sns.pairplot(data = dataset, hue = 'predict')
plt.savefig('pred.jpg')
plt.show()
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import confusion_matrix
from sklearn.model_selection import train_test_split
```

```python
x = dataset.iloc[:, :-1]
y = dataset['predict']
```

```python
x_train, x_test, y_train, y_test = train_test_split(x,y, test_size = 0.3, random_state = 42)
```

```python
lr = LogisticRegression()
lr.fit(x_train,y_train)
```

```python
lr.score(x_train, y_train)*100, lr.score(x_test, y_test)*100
```

```python
y_pred = lr.predict(x_test)
```

```python
cf = confusion_matrix(y_test, y_pred)
```

```python
sns.heatmap(cf, annot =True)
plt.show()
```

```python
lr.predict([[101.764052,100.160928,99.494642,99.909756,101.761266]])
```

```python
knn =  KNeighborsClassifier()
knn.fit(x_train, y_train)
```

```python
knn.score(x_train, y_train)*100, knn.score(x_test, y_test)*100
```

```python
knn.predict([[101.764052,100.160928,99.494642,99.909756,101.761266]])
```

```python
dtc = DecisionTreeClassifier()
dtc.fit(x_train, y_train)
```

```python
dtc.score(x_train, y_train)*100, dtc.score(x_test, y_test)*100
```

```python
from sklearn.ensemble import VotingClassifier
```

```python
li = [('lr',LogisticRegression()), ('knn',KNeighborsClassifier()), ('dts',DecisionTreeClassifier())]
```

```python
vc = VotingClassifier(li, weights = [10,5,7] )
vc.fit(x_train, y_train)
```

```python
vc.score(x_train, y_train)*100, vc.score(x_test, y_test)*100
```

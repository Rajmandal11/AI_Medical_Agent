```python
#VIF(The Variance Inflation Factor (VIF) is used to detect multicollinearity in multiple regression models by measuring how much the variance of a regression coefficient is inflated due to correlations between independent variables. A high VIF indicates that a predictor variable is redundant or strongly correlated with other predictors, which can lead to instability and unreliability in the regression model's coefficients.)
```

```python
from statsmodels.stats.outliers_influence import variance_inflation_factor

housing_fg_vif = housing_fg.drop("HousePrice", axis = 1)
housing_fg_vif = housing_fg_vif.apply(pd.to_numeric,errors = "coerce")

vif_data = pd.DataFrame()
vif_data["Feature"] = housing_fg_vif.columns
vif_data["VIF"] = [variance_inflation_factor(housing_fg_vif.values, i) for i in range (len(housing_fg_vif.columns))]
print(vif_data)
```

```python
housing_fg = housing_fg.drop(["Latitude","Longitude"],axis = 1)
```

```python
cols_to_convert = ['MedInc','HouseAge','AveRooms','AveBedrms','Population','AveOccup']
housing_fg[cols_to_convert] = housing_fg_vif[cols_to_convert].apply(pd.to_numeric,errors = "coerce")
housing_fg['HousePrice'] = pd.to_numeric(housing_fg['HousePrice'], errors = 'coerce')

x = sm.add_constant(housing_fg.drop('HousePrice', axis = 1))
y = housing_fg['HousePrice']

model = sm.OLS(y,x).fit()
```

```python
residuals = model.resid

plt.figure(figsize = (10,6))
plt.subplot(1,2,1)
sns.histplot(residuals, kde = True)
plt.title('Residuals distribution')

plt.subplot(1,2,2)
plt.scatter(model.predict(x), residuals)
plt.axhline(y=0, color = 'red', linestyle = '--')
plt.title("residuls vs Predicted")
plt.ylabel('Residuals')
plt.show()

print("Model coefficients:\n", model.params)
```

```python
from sklearn.metrics import mean_squared_error, r2_score

predictions = model.predict(x)
print("R squre Score: ", r2_score(y, predictions))
print("MSE: ", mean_squared_error(y, predictions))
```

```python
def plot_boxplots(fg):
    fig, axes = plt.subplots(1,len(fg.columns), figsize=(20,10))
    for i, column_name in enumerate(fg.columns):
        fg.boxplot(column=column_name, ax=axes[i])
        axes[i].set_title(column_name)
    plt.show()
plot_boxplots(housing_fg)
```

```python
def remove_outliers(fg):
    fg_no_outliners = fg.copy()

    for column_name in fg.columns:
        Q1 = fg[column_name].quantile(0.25)
        Q3 = fg[column_name].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    fg_no_outliners[column_name] = fg_no_outliners[column_name].apply(lambda x: lower_bound if x < lower_bound
                                                                   else upper_bound if x > upper_bound else x)
    return fg_no_outliners
housing_fg_new = remove_outliers(housing_fg)
```

```python
plot_boxplots(housing_fg_new)
```

```python
cols_to_convert = ['MedInc','HouseAge','AveRooms','AveBedrms','Population','AveOccup']
housing_fg_new[cols_to_convert] = housing_fg_new[cols_to_convert].apply(pd.to_numeric,errors = "coerce")
housing_fg_new['HousePrice'] = pd.to_numeric(housing_fg_new['HousePrice'], errors = 'coerce')

x = sm.add_constant(housing_fg_new.drop('HousePrice', axis = 1))
y = housing_fg_new['HousePrice']

model = sm.OLS(y,x).fit()
```

```python
residuals = model.resid

plt.figure(figsize = (10,6))
plt.subplot(1,2,1)
sns.histplot(residuals, kde = True)
plt.title('Residuals distribution')

plt.subplot(1,2,2)
plt.scatter(model.predict(x), residuals)
plt.axhline(y=0, color = 'red', linestyle = '--')
plt.title("residuls vs Predicted")
plt.xlabel("Predicted Values")
plt.ylabel('Residua')
plt.show()

print("Model coefficients:\n", model.params)
```

```python
from sklearn.metrics import mean_squared_error, r2_score

predictions = model.predict(x)
print("R squre Score: ", r2_score(y, predictions))
print("MSE: ", mean_squared_error(y, predictions))
```

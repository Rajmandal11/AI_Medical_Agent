```python
from scipy import stats
import numpy as np
import matplotlib.pyplot as plt

np.set_printoptions(precision = 6)

t_dist = stats.t(20)

t_values = np.linspace(-3, 4, 1000)

plt.plot(t_values,t_dist.pdf(t_values))
plt.xlabel("t value")
plt.ylabel("probability for t value")
plt.title("PDF for t distribution with df=20")

plt.plot(t_values,t_dist.cdf(t_values))
plt.xlabel("t value")
plt.ylabel("probability for t value <= t'")
plt.title("CDF for t distribution with df=20")

plt.show()
```

```python
def mean(arr):  
    res = sum(arr) / len(arr)  
    return res  

def sd(arr, n):  
    sum_sq_diff = 0  
    for i in range(n):  
        sum_sq_diff += (arr[i] - mean(arr)) ** 2  
    sdd = sum_sq_diff / n 
    return sdd  

def combinedVariance(A1, A2):  
    n = len(A1)  
    m = len(A2)  
    mean1 = mean(A1)  
    mean2 = mean(A2)  
    print("mean_1: ", round(mean1, 2), " mean_2: ", round(mean2, 2)) 

    sd1 = sd(A1, n)  
    sd2 = sd(A2, m)  
    print("sd_1: ", round(sd1, 2), " sd_2: ", round(sd2, 2))  

    combinedMean = (n * mean1 + m * mean2) / (n + m)  
    print("Combined Mean: ", round(combinedMean, 2))  

    d1_square = (mean1 - combinedMean) ** 2  
    d2_square = (mean2 - combinedMean) ** 2  
    print("d1_square: ", round(d1_square, 2), " d2_square: ", round(d2_square, 2))  

    comb_var = (n * (sd1 + d1_square) + m * (sd2 + d2_square)) / (n + m)  
    print("Combined Variance: ", round(comb_var, 2))  

def coeffOfVariance(A1): 
    n = len(A1) 
    mean1 = mean(A1)  
    sd1 = sd(A1, n)  
    cov = mean1 / sd1  
    print("Coefficient of Variance: ", round(cov, 2)) 

A1 = [24, 46, 35, 79, 13, 77, 35] 
A2 = [66, 68, 35, 24, 46] 
combinedVariance(A1, A2)  
coeffOfVariance(A1)  
```

```python
import numpy as np 
from sklearn.datasets import load_iris 
from sklearn.model_selection import train_test_split 
from sklearn.neighbors import KNeighborsClassifier 
from sklearn.metrics import accuracy_score 
from sklearn.preprocessing import StandardScaler 

 

# Load Iris dataset 
iris = load_iris() 
X = iris.data      # Features 
y = iris.target    # Class labels 

#Feature Scaling (Important for KNN) 
scaler = StandardScaler() 
X_scaled = scaler.fit_transform(X) 

#Split Dataset into Training and Testing 
X_train, X_test, y_train, y_test = train_test_split( X_scaled, y, test_size=0.2,random_state=42) 

#Train KNN Model for Different k Values 
k_values = [1, 3, 5, 8, 9, 11 ,15] 
accuracy_results = {} 

for k in k_values: 
    knn = KNeighborsClassifier(n_neighbors=k) 
    knn.fit(X_train, y_train) 
    y_pred = knn.predict(X_test) 
    accuracy = accuracy_score(y_test, y_pred) 
    accuracy_results[k] = accuracy 

    print(f"Accuracy for k = {k} : {accuracy:.2f}") 
    print("\n--- Accuracy Comparison for Different k Values ---") 
    
    for k, acc in accuracy_results.items(): 
        print(f"k = {k} → Accuracy = {acc:.2f}") 
```

```python
from sklearn.svm import SVC 

svm_linear = SVC(kernel='linear') 
svm_linear.fit(X_train, y_train) 

y_pred_linear = svm_linear.predict(X_test) 
accuracy_linear = accuracy_score(y_test, y_pred_linear) 
print("Accuracy using Linear Kernel:", accuracy_linear) 

 

#Polynomial Kernel 
svm_poly = SVC(kernel='poly', degree=3) 
svm_poly.fit(X_train, y_train) 

y_pred_poly = svm_poly.predict(X_test) 
accuracy_poly = accuracy_score(y_test, y_pred_poly) 
print("Accuracy using Polynomial Kernel:", accuracy_poly) 

#RBF Kernel 

svm_rbf = SVC(kernel='rbf') 
svm_rbf.fit(X_train, y_train) 

y_pred_rbf = svm_rbf.predict(X_test) 
accuracy_rbf = accuracy_score(y_test, y_pred_rbf) 
print("Accuracy using RBF Kernel:", accuracy_rbf) 
```

```python
import warnings 
warnings.filterwarnings("ignore", category=RuntimeWarning) 
from sklearn.cluster import KMeans 

# Load Iris dataset 
iris = load_iris() 
X = iris.data 

#Feature Scaling 
scaler = StandardScaler() 
X_scaled = scaler.fit_transform(X) 
k_values = [1,2, 3, 5, 7, 9, 12, 15] 
inertia_values = {} 

 

#Apply K-Means for Different k Values 
for k in k_values: 
    kmeans = KMeans(n_clusters=k, random_state=39) 
    kmeans.fit(X_scaled) 

    inertia_values[k] = kmeans.inertia_ 
    print(f"k = {k} → WCSS (Inertia) = {kmeans.inertia_:.2f}") 
    
# Display cluster labels for k = 3 
kmeans_3 = KMeans(n_clusters=3, random_state=42) 
labels = kmeans_3.fit_predict(X_scaled) 
print("\nCluster labels for k = 3:") 
print(labels) 
```

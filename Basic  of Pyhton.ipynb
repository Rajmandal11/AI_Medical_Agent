```python
# concepts of Inheritannce 
class GrandFather:
    def __init__(self,lname):
        self.lname = lname

    def GF_methods(self):
        print(f"This is methods from the GrandFather class: {self.lname}")

class Father(GrandFather):
    
    def __init__(self,address,money):
        self.address = address
        self.money = money
        
    def F_methods(self):
        print(f"This is methods from the Father class: {self.address}")
        print(f"This is methods from the Father class: {self.money}")

class Son(Father):
    def __init__(self,fname,age):
        self.fname = fname
        self.age = age
        
    def Son_methods(self):
        print(f"This is methods from the Son class: {self.fname}")
        print(f"This is methods from the Son class: {self.age}")

    def F_methods(self):
        print(f"This is methods from the Father class: {self.address}")
        print(f"This is methods from the Father class: {self.money}")
        
    def GF_methods(self):
        print(f"This is methods from the GrandFather class: {self.lname}")

gf = GrandFather("Mandal")
gf.GF_methods

f = Father("Kushmahar", None)
f.F_methods

s = Son("Raj",20)
s.Son_methods

print(gf.GF_methods())
print(f.F_methods())
print(s.Son_methods())
```

```python
# concepts of Inheritannce 

class Father:
    def __init__(self, lname):
        self.lname = lname

    def father_method(self):
        print(f'This is a method from the  father class. Attribute: {self.lname}')

class Son(Father):
    def __init__(self,fname, age):
        self.fname = fname
        self.age = age
        
    def son_method(self):
        print(f"This is a method from the son class. Attribute: {self.fname}")
        print(f"This is a method from the son class. Attribute: {self.age}")

    def father_method(self):
        print(f'This is overraide from the  son class. Attribute: {self.lname}')
        
f1 = Father("Mandal")
f1.father_method

s = Son("Raj", 20)
s.son_method
s.father_method

print(s.son_method(),"\n",f1.father_method())
```

```python
# polymorphism

class Animal:
    def make_sound(self):
        raise NotImplementedError("Subclasses must implement this method")

class Cow(Animal):
    def eat(self):
        return "gress"

class Dog(Animal):
    def make_sound(self):
        return  "Woow!"
        
class Cat(Animal):
    def make_sound(self):
        return "Quack!"
        
class Duck(Animal):
    def make_sound(self):
        return "Quack!"

c = Cow()
D = Dog()
Ct = Cat()
Dk = Duck()

print(c.eat())
print(D.make_sound())
print(Ct.make_sound())
print(Dk.make_sound())
```

```python
#Encaplication
class Employee:
    
    def __init__(self, name="Kal", age=32):
        self.__name = name  # Private attribute
        self.__age = age    # Private attribute

    def get_name(self):
        return self.__name

    def set_name(self, name):
        self.__name = name
    
    def get_age(self):
        return self.__age
    
    def set_age(self, age):
        self.__age = age 
    
# Create an instance of Employee
e1 = Employee()

# Accessing the name and age using getter methods
print(e1.get_name())
print(e1.get_age())
```

```python
rows = int(input("Enter the row size for the pattern: "))
for i in range(1, rows + 1):  # Upper part of diamond
    for j in range(rows - i):
        print(" ", end=" ")
    for k in range(1, 2 * i):
        if k == 1 or k == 2 * i - 1:  # Print star on borders
            print("*", end=" ")
        else:
            print(" ", end=" ")  # Print space inside
    print()
for i in range(rows - 1, 0, -1):  # Lower part of diamond
    for j in range(rows - i):
        print(" ", end=" ")
    for k in range(1, 2 * i):
        if k == 1 or k == 2 * i - 1:  # Print star on borders
            print("*", end=" ")
        else:
            print(" ", end=" ")  # Print space inside
    print()
```

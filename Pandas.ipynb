```python
# Merging & Joining
# SYNTAX
# pd.marge(au1, au2, on="Cloumn_Name", how="type of join")
import pandas as pd

#customer dataframe
d_customers = pd.DataFrame({
    "CustomerID" : [1,2,3],
    "Name": ['Rahul','Kamlesh','Manish']
})

# order dataframe
d_orders = pd.DataFrame({
    "CustomerID" : [1,2,4],
    "Order": [234,543,343]
})

d_merged = pd.merge(d_customers, d_orders, on="CustomerID", how="inner")
print("inner join")
print(d_merged)

d_merged = pd.merge(d_customers, d_orders, on="CustomerID", how="outer")
print(d_merged)

d_merged = pd.merge(d_customers, d_orders, on="CustomerID", how="left")
print(d_merged)

d_merged = pd.merge(d_customers, d_orders, on="CustomerID", how="right")
print(d_merged)

d_merged = pd.merge(d_customers, d_orders, on="CustomerID", how="cross")
print(d_merged)
```

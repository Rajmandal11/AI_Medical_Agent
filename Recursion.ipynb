```python
# Factorial
def fact(n):
    if(n == 1):
        return 1
    else:
        return n * fact(n - 1)
res = fact(6)
print(res)
```

```python
# Find number is odd or even
def val(n):
    if (n % 2 == 0):
        print('Prime number')
    else:
        print('Not prime number')
res = val(7)
```

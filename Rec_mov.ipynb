```python
moviess = movie[['movie_id','genres','cast','crew','title','keywords','overview']]
moviess.head(2)
```

```python
moviess.info()
```

```python
moviess.isnull().sum()
```

```python
moviess.dropna(inplace = True)
```

```python
moviess.iloc[0].genres
```

```python
import ast
ast.literal_eval('[{"id": 28, "name": "Action"}, {"id": 12, "name": "Adventure"}, {"id": 14, "name": "Fantasy"}, {"id": 878, "name": "Science Fiction"}]')
```

```python
def convert(obj):
    lis = []
    for i in ast.literal_eval(obj):
        lis.append(i['name'])
    return lis
```

```python
moviess['genres'].apply(convert)
```

```python
moviess['genres'] = moviess['genres'].apply(convert)
```

```python
moviess['keywords'] = moviess['keywords'].apply(convert)
```

```python
def convert1(obj):
    lis = []
    counter = 0
    for i in ast.literal_eval(obj):
        if counter != 3:
            lis.append(i['name'])
            counter+=1
        else:
            break
    return lis
```

```python
moviess['cast'] = moviess['cast'].apply(convert1)
```

```python
moviess.head(1)
```

```python
def fetch_dir(obj):
    lis = []
    for i in ast.literal_eval(obj):
        if i['job'] == 'Director':
            lis.append(i['name'])
            break
    return lis
```

```python
moviess['crew'] = moviess['crew'].apply(fetch_dir)
```

```python
moviess.head(1)
```

```python
moviess['overview'].apply(lambda x:x.split())
```

```python
moviess['overview'] = moviess['overview'].apply(lambda x:x.split())
```

```python
moviess.head(1)
```

```python
moviess['crew'].apply(lambda x:[i.replace(' ','') for i in x])
```

```python
moviess['crew'] = moviess['crew'].apply(lambda x:[i.replace(' ','') for i in x])
moviess['genres'] = moviess['genres'].apply(lambda x:[i.replace(' ','') for i in x])
moviess['cast'] = moviess['cast'].apply(lambda x:[i.replace(' ','') for i in x])
moviess['keywords'] = moviess['keywords'].apply(lambda x:[i.replace(' ','') for i in x])
```

```python
moviess.head(1)
```

```python
moviess['tags'] = moviess['crew'] + moviess['genres'] + moviess['cast'] + moviess['keywords']
```

```python
moviess.head(1)
```

```python
data = moviess[['movie_id','title','tags']]
data.head()
```

```python
data['tags'].apply(lambda x: ' '.join(x))
```

```python
data['tags'] = data['tags'].apply(lambda x: ' '.join(x))
```

```python
data.head()
```

```python
from nltk.stem.porter import PorterStemmer
ps = PorterStemmer()
```

```python
def stem(text):
    w = []
    for i in text.split():
        w.append(ps.stem(i))
    return " ".join(w)
```

```python
data['tags'] = data['tags'].apply(stem)
```

```python
data['tags'] = data['tags'].apply(lambda x:x.lower())
```

```python
data.head()
```

```python
from sklearn.feature_extraction.text import CountVectorizer
cv = CountVectorizer(max_features = 4000, stop_words = 'english')
```

```python
cv.fit_transform(data['tags']).toarray().shape
```

```python
vector = cv.fit_transform(data['tags']).toarray()
```

```python
vector[0]
```

```python
cv.get_feature_names_out()
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
similarity  = cosine_similarity(vector)
```

```python
sorted(list(enumerate(similarity[0])), reverse = True, key = lambda x:x[1])[1:6]
```

```python
def recommend(movie):
    movie_index = data[data['title'] == movie].index[0]
    distance = similarity[ movie_index]
    movie_list = sorted(list(enumerate(distance)), reverse = True, key = lambda x:x[1])[1:6]

    for i in movie_list:
        print(data.iloc[i[0]].title)
```

```python
recommend('Avatar')
```
